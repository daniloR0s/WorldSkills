```python
#Convertendo datas com horarios aleatorios
import numpy as np
rng = np.random.default_rng(42)
def add_random_time(date_series):
    dt = pd.to_datetime(date_series, errors="coerce") #ve as datas e transforma as que sao invalidas em NaT, not a time
    horas = rng.integers(9, 18, size=len(dt)) # 9 a 17, gera de 9 a 17
    minutos = rng.integers(0, 60, size=len(dt)) #gera de 0 a 59
    return dt + pd.to_timedelta(horas, unit="h") + pd.to_timedelta(minutos, unit="m")#Calma, explicarei, so retorna a variavel hora e na unidade h e o mesmo no outro


df_orders["order_date"] = add_random_time(df_orders["order_date"])
df_members["join_date"] = add_random_time(df_members["join_date"])

import re
def clean_phone(p):
    if p in (None, "0") or pd.isna(p):#se p(que é o numero de telefone), for none, 0 ou se for Is Not a, ou seja, nao for o padrao, retorna o normal, o que ja esta ali
        return p
    return re.sub(r"[^\d+]", "", str(p))#se for, a func re.sub vai padrão, substituiição, texto, no caso o padr~]ao é o [^\d+], e vamos substituir por ' '(espaço) onde
    #str(p), transforma o numero em string, re.sub é regex.sub que faz Quando ^ aparece logo depois do [, significa negação. significa qualquer caractere que NÃO seja um dígito ou +.
    #\d significa qualquer dígito de 0 a 9.[\d+] significa qualquer coisa que NÃO seja dígito nem +. ou seja encontre caracteres que não sejam números nem + e substitua por nada.
df_members["phone_number"] = df_members["phone_number"].apply(clean_phone)

ondeAusentes = df_members['age'].isnull().index.to_numpy()
for i in ondeAusentes:
    print(f'tem valor ausente em age na linha {i}')

#Nenhum horário está fixo em 00:00:00 (verificado com .dt.time.value_counts())

qtdMeiaNoite = (df_orders[df_orders['order_date'].dt.hour == 0]['order_date'].dt.time.value_counts())#dt.hour pega so a hora de um datetime

import os

os.makedirs(r"C:\GameStoreBrasil\Output", exist_ok=True)

df_members.to_csv("output/members_cleaned.csv", index=False)
df_orders.to_csv("output/orders_cleaned.csv", index=False)#criando um arquivo com pandas

def adicionarRandomTempo(date_series):
    #transformar em datas no padrao, se nao for, ele vira NaT
    dt = pd.to_datetime(date_series,errors='coerce')
    #gerar horas aleatorias de 9 a 17
    horas = rng.integers(9, 18, size = len(dt))
    #gerar minutos aleatorios de 0 a 59
    minutos = rng.integers(0,60, size = len(dt))
    
    return dt + pd.to_timedelta(horas, unit='h') + pd.to_timedelta(minutos, unit='m')
```
