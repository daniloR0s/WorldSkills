```python
#==============================================

#Tarefa 1.1, Carregamento e exploração de dados

#GameStore

#==============================================
import pandas as pd
import numpy as np
relatorio = []

print("="*60)
print("iniciando a tarefa 1.1 - GameStore Brasil")
print("="*60)
```

```python
#carregando os arquivos csv#

print("Carregando os arquivos da pasta 'data'...")
df_games = pd.read_csv('../data/games.csv')
df_orders = pd.read_csv('../data/orders.csv')
df_members = pd.read_csv('../data/members.csv')
print("Arquivos carregados com sucesso")
```

```python
#O comando .shape-> mostra quantas linhas e colunas o arquivo tem

print(f"orders.csv:{df_orders.shape[0]} linhas e {df_orders.shape[1]} colunas")
print(f"orders.csv:{df_members.shape[0]} linhas e {df_orders.shape[1]} colunas")
print(f"orders.csv:{df_games.shape[0]} linhas e {df_orders.shape[1]} colunas")
```

```python
#passo 3 -  verificar valores ausente(NaN)

print("\n - orders.csv(vendas) -- ")
print(df_orders.isnull().sum())
print("\n - members(Membros) -- ")
print(df_members.isnull().sum())
print("\n - games(Jogos)--")
print(df_games.isnull().sum())
```

```python
#Passo 4, exploração inicial (Primeiras Linhas)

print("\n orders.csv (Vendas)")

#mostra as 5 primeiras linhas do csv

print(df_orders.head())

print("\n games.csv (jogos)")
#mostra as 5 primeiras linhas do csv

print(df_games.head())

print("\n members.csv (membros)")
#mostra as 5 primeiras linhas do csv
print(df_members.head())
```

```python
#Passo 5 verificar os tipos de dados dtype
print("orders.csv (vendas)")
print(df_orders.dtypes)
print(df_orders.head())


print("members.csv (membros)")
print(df_members.dtypes)
print(df_members.head())


print("games.csv (games)")
print(df_games.dtypes)
```

```python
#Passo 6, detectar datas invalidas 

data_orders_invalidas = pd.to_datetime(df_orders['order_date'],errors="coerce").isna().sum()
print(f"datas invalidas em orders.csv (orders-date) {data_orders_invalidas}")

data_members_invalidas = pd.to_datetime(df_members['join_date'],errors="coerce").isna().sum()
print(f"datas invalidas em members.csv (join-date) {data_members_invalidas}")

data_games_invalidas = pd.to_datetime(df_games['release_date'],errors="coerce").isna().sum()
print(f"datas invalidas em games.csv (orders-date) {data_games_invalidas}")
```

```python
#Salvar no relatorio

relatorio.append(f"datas invalidas em orders.csv(orders-data) {data_orders_invalidas}")
relatorio.append(f"datas invalidas em members.csv(join_date) {data_members_invalidas}")
relatorio.append(f"datas invalidas em games.csv(release-date) {data_games_invalidas}")
print(relatorio[0])

#caçar valores negativos
#verificando a tabela de pedidos(orders.csv)
quant_negativa_orders = (df_orders["quantity"]<0).sum()
preco_negativo_orders = (df_orders["unit_price"]<0).sum()
print(f"quantidade negativo em orders:{quant_negativa_orders}")

preco_game_negativo = (df_games["price"]<0).sum()
custo_game_negativo = (df_games["cost"]<0).sum()

relatorio.append(f"precos negativos em orders.csv(unit_price) {preco_negativo_orders}")
relatorio.append(f"quantidade negativas em orders.csv (quantity) {quant_negativa_orders}")
relatorio.append(f"preco negativos em games.csv (price)  {preco_game_negativo}")
relatorio.append(f"custos negativos em games.csv (cost) {custo_game_negativo}" )

print(relatorio[3])
print(relatorio[4])
print(relatorio[5])
print(relatorio[6])
```
