```python
#Passo 5, geração de relatorio final PDF
import matplotlib as plt
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
```

```python

print("\n" + "=" * 60)
print("PASSO 5: GERANDO RELATORIO PDF (TAREFA 1.3)")
print('=' * 60)
```

```python
#1 Preparar os dados para agregação mensal
#Garantir que a coluna de data seja datatime e seja o indice 

df_orders_agg = df_orders.copy()

df_orders_agg['order_date'] = pd.to_datetime(df_orders_agg['order_date'])
```

```python
mensal = df_orders_agg.set_index('order_date').resample("ME").agg(
    receita_total=("total_venda", "sum"),
    ticket_medio=("total_venda", "mean"),
    numero_de_transacoes=("total_venda", "count")
).reset_index()
print(mensal)
```

```python
#Formatar o mes para exibição legivel ex('2024-01)

#Identificar o top3 meses por receita

mensal['mes_ano'] = mensal['order_date'].dt.strftime('%Y-%m')
top3 = mensal.nlargest(3,'receita_total')[['mes_ano', 'receita_total']]#Imprime os top 3
```

```python
#Top 3 meses

print('\n --- Top 3 Meses por Receita Total ---')
print(top3.to_string(index = False))
```

```python
#Gerar o PDF com 3 graficos de linha e 1 tabela

caminho_pdf = r'C:\GameStoreBrasil\output\Session1_SalesTrends_GameStore.pdf'
```

```python
with PdfPages(caminho_pdf) as pdf:
   
    # --- Pagina 1: Grafico de Receita Total ---
    fig, ax = plt.subplots(figsize=(10, 5))
    mensal.plot(x='mes_ano', y='receita_total', ax=ax, marker='o', color='#1f77b4', legend=False)
    ax.set_title('Receita Total por Mes')
    ax.set_ylabel('Receita (R$)') 
    ax.set_xlabel('Mes')
    plt.xticks(rotation=45)
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

    # --- Pagina 2: Grafico de Numero de Transacoes ---
    fig, ax = plt.subplots(figsize=(10, 5))
    mensal.plot(x='mes_ano', y='numero_de_transacoes', ax=ax, marker='s', color='#2ca02c', legend=False)
    ax.set_title('Numero de Transacoes por Mes')
    ax.set_ylabel('Quantidade')
    ax.set_xlabel('Mes')
    plt.xticks(rotation=45)
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

    # --- Pagina 3: Grafico de Ticket Medio ---
    fig, ax = plt.subplots(figsize=(10, 5))
    mensal.plot(x='mes_ano', y='ticket_medio', ax=ax, marker='^', color='#d62728', legend=False)
    ax.set_title('Ticket Medio por Mes')
    ax.set_ylabel('Ticket Medio (R$)')
    ax.set_xlabel('Mes')
    plt.xticks(rotation=45)
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

    # --- Pagina 4: Tabela Top 3 Meses ---
    fig, ax = plt.subplots(figsize=(6, 3))
    ax.axis('off') # Esconder os eixos do grafico, queremos apenas a tabela
   
    ax.set_title('Top 3 Meses por Receita Total', fontsize=5, fontweight='bold', pad=20)
   
    # Criar a tabela visual
    tabela = ax.table(
        cellText=top3.values,
        colLabels=['Mes', 'Receita Total (R$)'],
        cellLoc='center',
        loc='center'
    )
    tabela.scale(1.2, 1.5) # Aumentar o tamanho das celulas
    tabela.auto_set_font_size(False)
    tabela.set_fontsize(12)
   
    pdf.savefig(fig)
    plt.close(fig)

print(f"\nRelatorio PDF gerado com sucesso em: {caminho_pdf}")
print("TAREFA 1.3 CONCLUIDA COM SUCESSO!")
```
