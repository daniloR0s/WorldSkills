```python
#==============================================

#Tarefa 1.1, Carregamento e exploração de dados

#GameStore

#==============================================
import pandas as pd
import numpy as np
import re as re
relatorio = []

print("="*60)
print("iniciando a tarefa 1.3 - limpeza de dados")
print("="*60)
print("\n carregando os arquivos csv...")
```

```python
#carregando os arquivos csv#

print("Carregando os arquivos da pasta 'data'...\n")
df_games = pd.read_csv('../data/games.csv')
df_orders = pd.read_csv('../data/orders.csv')

print(f"-orders.csv, carregando, e tem {df_orders.shape[0]} linhas")

df_members = pd.read_csv('../data/members.csv')

print(f"members.csv carregado e tem {df_members.shape[0]} linhas e {df_members.shape[1]} colunas\n")
print("Arquivos carregados com sucesso")
```

```python
#Preenchendo com mediana nos valores ausentes para normalizar tudo

print(df_members['age'].isnull().sum())
mediana_age = df_members['age'].median()
df_members['age'] = df_members['age'].fillna(mediana_age)
print(f'mediana da idade calculada: {mediana_age}')

#preenchendo com 0 os valores ausentes em phone number e promotion id

df_members['phone_number'] = df_members['phone_number'].fillna("0")#fillna preenche os valores nulos na coluna com o q eu quiser
print(f'telefones ausentes apos o preenchimento {df_members['phone_number'].isnull().sum()}')

df_orders['promotion_id'] = df_orders['promotion_id'].fillna("0")
print(f'Promoções ausentes apos o preenchimento {df_orders['promotion_id'].isnull().sum()}')


```

```python
#Verificar se ainda existem valores ausentes na coluna age 

ausentes_restantes = df_members['age'].isnull().sum()
print(f'Valores ausentes em age apos o preenchimento {ausentes_restantes}')
```

```python
#Verificar a questão do phone number, trocar ne
def limparCell(p):
    if p in(None, 0) or pd.isna(p):
        return p
    return re.sub(r"[^\d+]", " ", str(p))
rng = np.random.default_rng(42)
def adicionarRandomTempo(date_series):
    #transformar em datas no padrao, se nao for, ele vira NaT
    dt = pd.to_datetime(date_series,errors='coerce')
    #gerar horas aleatorias de 9 a 17
    horas = rng.integers(9, 18, size = len(dt))
    #gerar minutos aleatorios de 0 a 59
    minutos = rng.integers(0,60, size = len(dt))
    
    return dt + pd.to_timedelta(horas, unit='h') + pd.to_timedelta(minutos, unit='m')
```

```python
#Aplicando nos dataframes

df_members["phone_number"] = df_members["phone_number"].apply(limparCell)
df_members["join_date"] = df_members["join_date"].apply(adicionarRandomTempo)
df_orders["order_date"] = df_orders["order_date"].apply(adicionarRandomTempo)

#Pode ser também

df_orders['order_date'] = adicionarRandomTempo(df_orders['order_date'])
df_members['join_date'] = adicionarRandomTempo(df_members['join_date'])
```
